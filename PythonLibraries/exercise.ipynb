```python
import numpy as np;
sudoku = np.array([[5,3,4,6,7,8,9,1,2],
                   [6,7,2,1,9,5,3,4,8],
                   [1,9,8,3,4,2,5,6,7],
                   
                   [8,5,9,7,6,1,4,2,3],
                   [4,2,6,8,5,3,7,9,1],
                   [7,1,3,9,2,4,8,5,6],

                   [9,6,1,5,3,7,2,8,4],
                   [2,8,7,4,1,9,6,3,5],
                   [3,4,5,2,8,6,1,7,9]
                  ])
sudoku
```

```python
b=np.sum(sudoku,axis=0)
for i in b:
    if i !=45:
        print("suduku is not valid")
        break
    else:
        print("for rows it is valid")
```

```python
c=np.sum(sudoku,axis=1)
for i in c:
    if i !=45:
        print("sudoku is not valid")
        break
    else:
        print("for columns it is valid")
```

```python
slice = sudoku[0:3,0:3]
print(slice)
np.sum(slice,axis=0)
```

```python
# slicing all the arrays at once using loops
for i in range(0,9,3):
    for j in range(0,9,3):
        n=sudoku[i:i+3,j:j+3]
        print(n)
        print(n.sum())
        print(f"{i}sss{j}")
```

```python
# column:[age,math-marks,science-marks]
import numpy as np;
data = np.array([[18,85,78],
                 [19,92,88],
                 [17,76,95],
                 [18,65,70],
                 [20,90,85],
                ])
data
```

```python
# find the average of student
np.mean(data[:,0])
```

```python
# extract math marks of all students.
data[:,1]
```

```python
# find the highest science mark;
np.max(data[:,2])
```

```python
# get details of the student who scored more than 90 in math.
data[:,1]>90
```

```python
# increase math marks of all students by 5

data[:,1]+=5
data
```

```python
# find how many students are younger than 19
len(data[data[:,0]<19])
```

```python
# calculate the average marks in each subject (column-wise mean).
np.mean(data[:,1:],axis=0)
```

```python
# get data of students who scored at least 80 in both subjects.

data[(data[:,1]>=80) & (data[:,2]>=80)]

```

```python
import numpy as np

# Your 9x9 array
arr = np.array([
    [5, 3, 4, 6, 7, 8, 9, 1, 2],
    [6, 7, 2, 1, 9, 5, 3, 4, 8],
    [1, 9, 8, 3, 4, 2, 5, 6, 7],
    [8, 5, 9, 7, 6, 1, 4, 2, 3],
    [4, 2, 6, 8, 5, 3, 7, 9, 1],
    [7, 1, 3, 9, 2, 4, 8, 5, 6],
    [9, 6, 1, 5, 3, 7, 2, 8, 4],
    [2, 8, 7, 4, 1, 9, 6, 3, 5],
    [3, 4, 5, 2, 8, 6, 1, 7, 9]
])

# 1. Print each row separately
print("=== ROWS ===")
for i in range(arr.shape[0]):
    print(f"Row {i+1}: {arr[i]}")

# 2. Print each column separately
print("\n=== COLUMNS ===")
for j in range(arr.shape[1]):
    print(f"Column {j+1}: {arr[:, j]}")

# 3. Print the full array
print("\n=== FULL ARRAY ===")
print(arr)
```
