```python
import numpy as np
import pandas as pd
my_list =[10,20,30]
arr = np.array([10,20,30])
d = {"a":10,2:20,3:30}
```

```python
pd.Series(my_list)
# in pandas indexing is called "labels" and we can set custom labels to it.
```

```python
# setting custom labels
labels=["a","b","c"]
pd.Series(my_list,index=labels)
```

```python
pd.Series(arr)
```

```python
# creating series using dictionaries(objects)
pd.Series(d)
```
