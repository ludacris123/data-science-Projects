```python
import numpy as np
import pandas as pd
```

```python
data ={
    "A":[1,2,np.nan,4,5],
    "B":[np.nan,2,3,4,5],
    "C":[1,2,3,np.nan,np.nan],
    "D":[1,np.nan,np.nan,np.nan,5],
}
df=pd.DataFrame(data)
df
```

```python
# below line shows how many nanvalue in data and it gives answer like column wise
df.isna()
# below value gives the sum of all the nan value in columns
df.isna().sum()
```

```python
# this will give the ansewer if any value in data columns is not presenting nan vlaue
df.isna().any()
```

```python

df
```

```python
# removeing missing data
df.dropna()
```

```python
# drop threshold function
df
df.dropna(thresh=3)
```

```python
# filling the missing data
# this is not gonna permanently change the data if you want to then use (inplace= truee)
df.fillna(0)
```

```python
# can fill value in specific column in specific value;
values = {"A":0,"B":100,"C":300,"D":400}
df.fillna(value=values)
```
