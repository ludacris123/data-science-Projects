```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0,5,11)
y=x**2
y
```

```python
plt.plot(x,y)
plt.title("hello") # this gives heading to the graph
plt.xlabel("Chacha")
plt.ylabel("chachi")
```

```python
plt.figure(figsize=(15,8),dpi=100)
plt.subplot(2,3,1)
plt.plot(x,y)
plt.subplot(2,3,4)
plt.plot(x,y)
plt.subplot(2,3,3)
plt.plot(y,x)
plt.subplot(2,3,2)
plt.plot(x**2,y)
```

```python
# object oriented way
fig = plt.figure(figsize=(15,8),dpi=100)# figsize is for setting lenght and widht of canvas and dpi is for image quality(pixels)
axis1=fig.add_axes([0.1,0.1,0.5,0.5])#first value move right ,second value is move left,third value width ,fourth value height
axis1.plot(x,y)
axis2=fig.add_axes([1,0.1,0.5,0.5])#if you plot two graphs in one canvas then you can move it with the right and left value
axis2.plot(x**2,y)
```

```python
# types of plots
plt.scatter(x,y)#use for scatter plots
```
