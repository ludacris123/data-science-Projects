```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# CATEGORICAL DATA PLOTS
df=sns.load_dataset('tips')
df
```

```python
# this helps to count data plots for categorical data only
sns.countplot(x=df['sex'])
sns.countplot(x=df['sex'],hue=df['smoker'])
```

```python
sns.barplot(x =df['sex'],y=df['total_bill'])
```

```python
sns.barplot(x=df['sex'],y=df['total_bill'],estimator=np.sum)
```

```python
# boxplot
sns.boxplot(x='tip',y="day",data=df,hue='day', palette='rainbow')
# boxplot and violinplot are used to shown the distribution of categorical data. A box plot(or box-and-whisker plot) shows the distribution of quantitative
# in a way that facilitates comparison between variables or across levels of a categorical variable. the box shows the qauriles of the dataset while the whiskers
# extend to show the rest o the distribution, except for points that are determined to be "outliers" using a method that is a function of the inter-quartile range.
```

```python
sns.violinplot(x='day',y ='total_bill',data=df)
```

```python
# STRIPPLOT AND SWARMPLOT --> the stripplot will draw a scatterplot where one variable is categorical . A strip plot can be drawn on its own, but it is 
# also good complement to a box or violin plot in cases where you want to show all observation along with some representation of the underlying distribution.

# the swarmplot si similar to stripplot(),but the points are adjusted(only along the categorical axix) so that they don't overlap. this gives a 
# better representation of the distribution of values,although it does not scale as well to large numbers of observations (both in terms of the )
# ability to showw all the points and in terms of the computation needed to arrange them.
```

```python
sns.stripplot(x='day',y='total_bill',data=df)
```

```python
sns.swarmplot(x='day', y='total_bill',data=df)
```

```python
# swarmplot and violinplot looks so similar so we can plot them togetther
sns.violinplot(x='tip',y='day',data=df)
sns.swarmplot(x='tip',y='day',data=df)
```
