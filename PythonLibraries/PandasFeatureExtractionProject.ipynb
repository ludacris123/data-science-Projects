```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_csv(r'anime.csv')
df.head()
```

```python
# Make a new column for episode count
def extract_Episode(txt):
    check = False
    data = ""
    for i in txt:
        if i == ")":
            check =False
            return data
        if check == True:
            data += i
        if i == "(":
            check =True
            
            
```

```python
df["Episodes"] = df["Title"].apply(extract_Episode)
df['Episodes']=df['Episodes'].str.replace('eps',"")
df['Episodes']=df['Episodes'].astype(int)
df['Title'][0]
```

```python
# make a new column for time stamps
def extraction_time(txt):
    check =False;
    data=""
    for i in range(len(txt)):
        if txt[i] == ")":
            for j in range(i+1,i+20):
                data += txt[j]
            return data
df["Total Time"]=df['Title'].apply(extraction_time)  
df.head()
```

```python
# extracting months
from dateutil.relativedelta import relativedelta
from datetime import datetime

def calculate_total_months(period):
    try:
        start_str,end_str = period.split(' - ')
        start_date = datetime.strptime(start_str,"%b %Y")
        end_date = datetime.strptime(end_str,"%b %Y")
        r = relativedelta(end_date,start_date)
        return r.years* 12+ r.months + 1 # +1 to include the starting months
    except:
        return None

df["Months"]= df['Total Time'].apply(calculate_total_months)  
df.head()
```
