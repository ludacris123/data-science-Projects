```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=sns.load_dataset('tips')
df
```

```python
df['size'].unique()
```

```python
# we can only show qauntitative data here . note that all numbers are in datalist is not quantitative .
plt.subplot(1,2,1)
sns.histplot(df['total_bill'],bins=20,kde =True)# this is known as histogram (u can show this with also distplot)
# kde is known as kernel distributor show skewness in data.
plt.subplot(1,2,2)
sns.histplot(df['tip'],bins=20,kde =True)
```

```python
# joint plot
sns.jointplot(x="total_bill", y='tip', data= df,kind='hex')#kind : { "scatter" | "kde" | "hist" | "hex" | "reg" | "resid" }
```

```python
# sns.pairplot(df,hue='sex')#sex show male and female 
sns.pairplot(df,hue='size',palette="rainbow")
```

```python
# rugplot
# sns.rugplot(df['total_bill'])
sns.rugplot(df)
```
