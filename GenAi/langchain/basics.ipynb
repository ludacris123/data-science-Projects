```python
MODELS --> in practical Generative AI, there are mainly 3 important types of models:

chat/Language Models-->
1 Generate text
2. Answer questions
3. write code
4. summarize content

Embedding Models---->
1. convert text into numbers(vectors)
2. used in search and RAG
3. Help find similar content

Multimodal Models
1. work with more than just text
2. Can process images, audio or files
```

```python
PROMPTS--> A prompt is the instruction we give to a languge model.
it tells the models:
. what to do 
. how to respond
. what format to use
without prompts, the model has no direction.

* types of prompts
1. simple Prompt -> just a direct instruction
2. system + User Prompt--> modern chat models use roles: system-sets behavior and user -> ask question
3. prompt templates --> instead of writing prompts manually every time, we create reausable templates.
4. structured prompts --> used when we want JSON output , Bullet points , fixed format
```

```python
CHAINS --> a chain is a sequence of steps connected together.
instead of: one prompt--> one response we create: step1-> step2-->step3--> final output

Real AI application are not one-step.
EXAMPLE:
User asks:
summarize this ariticle and translate it into Hindi.
what actually happens?
1 Summarize text 2 Translate summary 3 Return result
That's a chain.
```

```python
MEMORY
Memory allows the ai system to remember past interactions.
without memory:
every message is treated independently.
with memory:
the model can use previous context.
    
```

```python
INDEXES --> indexes allow us to connect external data to an LLM.
LLMS only know what they were trained on. 
indexes help them use:
.pdfs
.documents
.databases
.company data
.website
LLMS: don't know your company data, private documents , don't update in real-time
so how do we solve this?
we connect external knowledge. that system is called "Retrieval"
```

```python
AGENTS--> agents are ai systems that can decide what action to take.
Unlike normal chains:
chains-> follow fixed steps
Agents-> Decide the next step dynamically
Normal LLM:
    user->Prompt-Response
but what if the task requires: searching the web, doing calculations, calling an api, using a database (the llm alone can't do it it need "tools")
```

```python
langchain_core.message
1. systemMessage is used to define how the ai should behave during the conversation. it sets the tone, personality and rule for the assistant
2. HUMANMESSAGE represent the input given by the user. whenever the user types something, it is wrapped inside a HumanMessage object.
3. AIMessage represent the response generated by the model. after the model replies, we store that reply as and AIMessage so it becomes part of the
conversation History. this helps maintain context in future interactions .
```

```python
# PROMPT TEMPLATE
from dotenv import load_dotenv
from langchain_core.prompts import ChatPromptTemplate
from pydantic import BaseModel
from typing import List,Optional
from langchain_core.output_parsers import PydanticOutputParser
# why we use pydanticOutputParser --> it acts like teacher (if student have to fill for and make mistake then it checks whether age is no. check box etc)

load_dotenv()
from langchain_mistralai import ChatMistralAI


model = ChatMistralAI(model = 'mistral-small-2506')



class Movie(BaseModel):
    title: str 
    release_year : Optional[int]
    genre: List[str]
    director: Optional[str]
    cast: List[str]
    rating: Optional[float]
    summary: str



parser = PydanticOutputParser(pydantic_object=Movie)


prompt = ChatPromptTemplate.from_messages([
    ('system',"""Extract movie information from the paragraph{format_instructions}"""),
    ("human","{paragraph}")
])
 
para = input("Give your paragraph : ")

final_prompt = prompt.invoke(
    {
        "paragraph" : para,
        'format_instructions': parser.get_format_instructions()
     }
)

response = model.invoke(final_prompt)
movie_data = parser.parse(response.content)

print(movie_data)
```
