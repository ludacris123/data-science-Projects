```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt 
import ast
import nltk
import warnings
warnings.filterwarnings('ignore')
```

```python
movies = pd.read_csv('tmdb_5000_movies.csv')
credits = pd.read_csv('tmdb_5000_credits.csv')
movies.head(1)
```

```python
credits.head(1)
```

```python
movies=movies.merge(credits,on='title')
movies.head(1)
```

```python
movies = movies[['movie_id', 'title', 'overview', 'genres', 'keywords', 'cast', 'crew']]
movies.head()
```

```python
# now we have to merge the important things from 'genre','keyword','cast','crew' and make one tag with it 
movies.isnull().sum()
```

```python
movies.dropna(inplace=True)
movies.isnull().sum()
```

```python
movies.duplicated().sum()
```

```python
# now change the formats of the columns
movies.iloc[0].genres
```

```python
# '[{"id": 28, "name": "Action"}, {"id": 12, "name": "Adventure"}, {"id": 14, "name": "Fantasy"}, {"id": 878, "name": "Science Fiction"}]' we need to
# change this to this format['Action','Adventure','FFantasy','SciFi']

def convert(obj):
    L=[]
    for i in ast.literal_eval(obj): #this line convert that string into integer
        L.append(i['name']) 
    return L
```

```python
# there is a problem if we want to comvert that string list to this format then this throw error first we conveth in in integer

movies['genres']=movies['genres'].apply(convert)
movies.head(1)
```

```python
movies['keywords']=movies['keywords'].apply(convert)
movies.head(1)
```

```python
def convert2(obj):
    L=[]
    counter=0
    for i in ast.literal_eval(obj): #this line convert that string into integer
        if counter !=3:
            L.append(i['name']) 
            counter+=1
        else:
            break
    return L
```

```python
movies['cast']=movies['cast'].apply(convert2)
movies.head(1)
```

```python
def fetchDirector(obj):
    L=[]
    for i in ast.literal_eval(obj): #this line convert that string into integer
        if i['job']=='Director':
            L.append(i['name']) 
    return L
```

```python
movies['crew']=movies['crew'].apply(fetchDirector)
movies.head(1)
```

```python
movies['overview']=movies['overview'].apply(lambda x:x.split())
movies.head(1)
```

```python
# remove space between name like in name rishabh patel same guy name but with space so we remove it for better understanding for models
```

```python
movies['genres'] = movies['genres'].apply(lambda x:[i.replace(" ", "") for i in x])
movies['keywords'] = movies['keywords'].apply(lambda x:[i.replace(" ", "") for i in x])
movies['cast'] = movies['cast'].apply(lambda x:[i.replace(" ", "") for i in x])
movies['crew'] = movies['crew'].apply(lambda x:[i.replace(" ", "") for i in x])
movies.head(1)
```

```python
movies['tags']= movies['overview'] + movies['genres'] + movies['keywords'] + movies['cast'] + movies['crew'] 
movies.head(1)
```

```python
new_df = movies[['movie_id','title','tags']]
new_df.head(1)
```

```python
new_df['tags']=new_df['tags'].apply(lambda x:" ".join(x))
new_df.head(1)
```

```python
new_df['tags']=new_df['tags'].apply(lambda x:x.lower())
new_df.head(1)
```

```python
from nltk.stem.porter import PorterStemmer
ps =  PorterStemmer()
```

```python
def stem (text):
    y = []
    for i  in text.split():
        y.append(ps.stem(i))
    return " ".join(y)
```

```python
new_df['tags']=new_df['tags'].apply(stem)
```

```python
# vectorization (now do ml)
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=5000,stop_words='english')
```

```python
vectors = cv.fit_transform(new_df['tags']).toarray()
vectors
```

```python
pd.DataFrame(cv.get_feature_names_out())
```

```python
from sklearn.metrics.pairwise import cosine_similarity
similarity = cosine_similarity(vectors)
similarity[1]
```

```python
def recommend(movie):
    movie_index = new_df[new_df['title']==movie].index[0]
    distances = similarity[movie_index]
    movies_list = sorted(list(enumerate(distances)),reverse=True,key=lambda x:x[1])[1:6]
    for i in movies_list:
        print(new_df.iloc[i[0]].title)

```

```python
recommend('Avatar')
```
