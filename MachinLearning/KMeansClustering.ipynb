```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
import warnings
warnings.filterwarnings('ignore')
```

```python
X, y = make_blobs(n_samples=500, centers=3, cluster_std=0.60, random_state=42)
y
```

```python
df= pd.DataFrame(X,columns=['Features1','Features2'])
df
```

```python
scaler = StandardScaler()
x_scaled = scaler.fit_transform(df)
```

```python
# now create elbow method to find k value

inertia=[]
k_range = range(1,11)
```

```python
for k in k_range:
    Kmeans = KMeans(n_clusters=k,random_state=42)
    Kmeans.fit(x_scaled)
    inertia.append(Kmeans.inertia_)
inertia
```

```python
plt.plot(k_range,inertia,marker ='o') 
```

```python
kmeans_final = KMeans(n_clusters=3,random_state=42)
```

```python
cluster_labels = kmeans_final.fit_predict(x_scaled)
cluster_labels
```

```python
df['cluster']=cluster_labels
```

```python
sns.scatterplot(x=df['Features1'],y=df['Features2'],hue=df['cluster'],palette='viridis')
```
